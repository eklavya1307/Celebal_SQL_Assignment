```python
pd.read_sql(
"""
SELECT COUNT(*) AS Total_Rows
FROM superstore
""",
engine
)
```

```python
# Display First 10 Rows
pd.read_sql("""
SELECT *
FROM superstore
LIMIT 10
""", engine)
```

```python
# Total Records
pd.read_sql("""
SELECT COUNT(*) AS Total_Rows
FROM superstore
""", engine)
```

```python
#Orders from West Region
pd.read_sql("""
SELECT *
FROM superstore
WHERE Region='West'
""", engine)
```

```python
# Technology Criteria
pd.read_sql("""
SELECT *
FROM superstore
WHERE Category='Technology'
""", engine)
```

```python
#Sales Greater Than 1000
pd.read_sql("""
SELECT *
FROM superstore
WHERE Sales > 1000
""", engine)
```

```python
# Total Sales
pd.read_sql("""
SELECT SUM(Sales) AS Total_Sales
FROM superstore
""", engine)
```

```python
# Total Profit
pd.read_sql("""
SELECT SUM(Profit) AS Total_Profit
FROM superstore
""", engine)
```

```python
#Sales by Category
pd.read_sql("""
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
ORDER BY Total_Sales DESC
""", engine)
```

```python
# Profit by Category
pd.read_sql("""
SELECT Category,
       SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit DESC
""", engine)
```

```python
# Sales by Region
pd.read_sql("""
SELECT Region,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Region
ORDER BY Total_Sales DESC
""", engine)
```

```python
# Top 10 Products
pd.read_sql("""
SELECT `Product Name`,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY `Product Name`
ORDER BY Total_Sales DESC
LIMIT 10
""", engine)
```

```python
#Top 10 Customers
pd.read_sql("""
SELECT `Customer Name`,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY `Customer Name`
ORDER BY Total_Sales DESC
LIMIT 10
""", engine)
```

```python
#Monthly Sales Trend
pd.read_sql("""
SELECT
DATE_FORMAT(
STR_TO_DATE(`Order Date`, '%%c/%%e/%%Y'),
'%%Y-%%m'
) AS Month,
SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Month
ORDER BY Month
""", engine)
```

```python
#Monthly Profit Trend
pd.read_sql("""
SELECT
DATE_FORMAT(
STR_TO_DATE(`Order Date`, '%%c/%%e/%%Y'),
'%%Y-%%m'
) AS Month,
SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Month
ORDER BY Month
""", engine)
```

```python
#Most Profitable Category
pd.read_sql("""
SELECT Category,
SUM(Profit) AS Profit
FROM superstore
GROUP BY Category
ORDER BY Profit DESC
LIMIT 1
""", engine)
```

```python
#Top Customer by Sales
pd.read_sql("""
SELECT `Customer Name`,
SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY `Customer Name`
ORDER BY Total_Sales DESC
LIMIT 1
""", engine)
```

```python
#Duplicate Check
pd.read_sql("""
SELECT `Row ID`,
COUNT(*) AS Duplicate_Count
FROM superstore
GROUP BY `Row ID`
HAVING COUNT(*) > 1
""", engine)
```

```python
#NULL Value Check
pd.read_sql("""
SELECT *
FROM superstore
WHERE Sales IS NULL
OR Profit IS NULL
""", engine)
```

```python
#Final Row Count Validation
pd.read_sql("""
SELECT COUNT(*) AS Total_Rows
FROM superstore
""", engine)
```
